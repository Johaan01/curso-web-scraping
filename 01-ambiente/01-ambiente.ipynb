# Módulo 01: Preparando o ambiente

Este é o primeiro módulo do curso de **Web Scraping com Python**.

*Web scraping* é a técnica de escrever um programa que **visita páginas da internet e extrai os dados de interesse**, como preços, nomes, notas e endereços, organizando-os em uma tabela. O que levaria horas de cópia manual, página por página, passa a ser feito de forma automática e padronizada.

## Objetivo do curso

Ao final do curso, você terá construído um programa (um *scraper*) que percorre um site inteiro e transforma o que encontrou em um **dataset**, uma tabela de dados pronta para ser analisada:

<img src="imagens/destino-dataset.png" alt="Uma página de livros à esquerda e, à direita, a tabela de dados extraída dela" width="1140">

O curso é dividido em 10 módulos, publicados à medida que ficam prontos. Este é o primeiro:

<img src="imagens/mapa-do-curso.png" alt="Mapa dos 10 módulos do curso, com o Módulo 01 destacado" width="1100">

## Neste módulo você vai

- Abrir e usar o **terminal** do seu computador;
- Instalar o **Python 3.14**;
- Criar um **ambiente virtual** para o curso;
- Instalar as **bibliotecas** do curso com o `pip`;

# Parte 1: Instalando tudo

## 1.1 As peças do ambiente

Antes de instalar, vale entender o papel de cada peça. Cada seção desta Parte 1 trata de uma delas:

<img src="imagens/pecas-do-ambiente.png" alt="As quatro peças do ambiente: Python, ambiente virtual, pip e bibliotecas, Jupyter ou VS Code, e o terminal por trás de tudo" width="1100">

- **Python** é a linguagem de programação e também o programa que *executa* o código que escrevemos (o "interpretador").
- O **ambiente virtual** é uma pasta chamada `.venv` que guarda as bibliotecas **usadas neste curso**, isoladas do restante do computador.
- **Bibliotecas** são códigos prontos que adicionam funcionalidades ao Python: o `requests` baixa páginas da internet, o `parsel` localiza informações dentro delas e o `pandas` organiza os dados em tabelas. Elas são instaladas pelo **pip**, o gerenciador de pacotes do Python.
- O **Jupyter** e o **VS Code** são os programas em que o código é escrito e executado, em *notebooks* como este.

A instalação e a configuração de tudo isso são feitas pelo **terminal**.

## 1.2 O terminal

O terminal é um programa em que o computador é operado por **comandos digitados**, em vez de cliques. É a forma padrão de instalar e configurar ferramentas de programação, e neste curso serão necessários poucos comandos.

### Como abrir

| Sistema | Como abrir o terminal |
|---|---|
| **Windows 10/11** | Clique com o botão direito no botão **Iniciar** e escolha **Terminal** (ou **Windows PowerShell**). Também é possível pesquisar "Terminal" no menu Iniciar. |
| **macOS** | Aperte `Cmd ⌘ + Espaço`, digite **Terminal** e aperte `Enter`. |
| **Linux** | `Ctrl + Alt + T` na maioria das distribuições. |

### Lendo o terminal

<img src="imagens/terminal-anatomia.png" alt="Janelas de terminal do Windows e do macOS, com o prompt e o comando cd destacados" width="900">

- ① O começo da linha, chamado de **prompt**, mostra **em qual pasta você está**. Todo comando é executado "dentro" dessa pasta.
- ② Você digita um comando e aperta **`Enter`**. Aqui, `cd Documents` significa "entre na pasta Documents" (*cd* vem de *change directory*).
- ③ O prompt mudou: agora você está **dentro** de `Documents`.
- ④ O cursor piscando indica que o terminal está **esperando** o próximo comando.

### Os comandos que você precisa conhecer

| O que faz | Windows (PowerShell) | macOS / Linux |
|---|---|---|
| Entrar em uma pasta | `cd NomeDaPasta` | `cd NomeDaPasta` |
| Voltar para a pasta de cima | `cd ..` | `cd ..` |
| Listar o que tem na pasta atual | `dir` | `ls` |

> 💡 **Atalhos:** a tecla **`Tab`** completa nomes de pastas e arquivos (digite `cd Doc` e aperte `Tab`). A **seta para cima `↑`** traz de volta os comandos anteriores.

> 💡 **Os nomes no terminal nem sempre são os que aparecem na tela.** O Windows traduz alguns nomes apenas na exibição: a pasta mostrada como "Documentos" no Explorador de Arquivos se chama `Documents` no terminal, e a "Área de Trabalho" se chama `Desktop`. O nome da pasta do usuário também pode ser diferente do nome da conta, porque acentos e espaços costumam ser removidos ou trocados: um usuário "João" pode ter a pasta `C:\Users\joao` ou `C:\Users\joao~`, por exemplo. Na dúvida, use `dir` para ver os nomes reais, ou a tecla `Tab` para completá-los. O mesmo acontece no macOS.

Desde o Python 3.14, a forma recomendada de instalar o Python no Windows é o **Python install manager**, um programa que baixa e gerencia as versões do Python instaladas no computador.

## 1.6 Criando o ambiente virtual

### Por que isolar as bibliotecas?

Projetos diferentes costumam depender de versões diferentes das mesmas bibliotecas. Se tudo for instalado diretamente no Python do computador, atualizar uma biblioteca para um projeto pode quebrar outro. O **ambiente virtual** resolve isso dando a cada projeto o seu próprio conjunto de bibliotecas:

<img src="imagens/venv-por-que.png" alt="Comparação: sem ambiente virtual, dois projetos disputam a versão do pandas; com ambiente virtual, cada projeto tem a sua caixa" width="1060">

Na prática, o ambiente virtual é uma pasta chamada `.venv` dentro da pasta do curso, onde ficam todas as bibliotecas instaladas. Se algo der errado, basta apagar a pasta e criá-la de novo.

### Criando

Com o terminal **na pasta do curso** (seção 1.5), digite o comando do seu sistema:

| Sistema | ⌨️ Comando |
|---|---|
| Windows, se instalou pela seção 1.4 | `py -3.14 -m venv .venv` |
| Windows, se já tinha Python 3.11 a 3.14 | `python -m venv .venv` |
| macOS | `python3.14 -m venv .venv` |
| Linux | `python3 -m venv .venv` |

O que cada parte do comando significa:

<img src="imagens/comando-venv.png" alt="O comando py -3.14 -m venv .venv dividido em três partes: qual Python usar, o que fazer e onde criar" width="980">

O comando leva alguns segundos e **não exibe nenhuma mensagem** quando é bem-sucedido, o que é comum entre comandos de terminal. Para conferir, liste a pasta: deve aparecer uma pasta nova chamada `.venv`. No Windows, use `dir`; no macOS e no Linux, use `ls -a` (pastas cujo nome começa com ponto ficam escondidas no `ls` comum).

### Ativando

Depois de criado, o ambiente precisa ser **ativado** para que o terminal passe a usar o Python e as bibliotecas que estão dentro dele.

| Sistema / terminal | ⌨️ Comando para ativar |
|---|---|
| Windows: PowerShell ou Terminal | `.venv\Scripts\Activate.ps1` |
| Windows: Prompt de Comando (cmd) | `.venv\Scripts\activate.bat` |
| macOS / Linux | `source .venv/bin/activate` |

Quando a ativação funciona, aparece **`(.venv)`** no começo da linha do terminal:

<img src="imagens/terminal-venv-ativado.png" alt="Terminal mostrando a criação do ambiente, o comando de ativação e o (.venv) no começo da linha" width="1000">

- ① O comando que cria o ambiente. A ausência de mensagem indica que deu certo.
- ② O comando que ativa o ambiente.
- ③ A confirmação da ativação: **`(.venv)`** no começo da linha.

> ⚠️ **Erro comum no Windows:** se aparecer uma mensagem em vermelho dizendo que a *"execução de scripts foi desabilitada neste sistema"*, o Windows está bloqueando scripts por segurança. Veja na imagem abaixo como liberar **apenas para o seu usuário**:

<img src="imagens/terminal-erro-scripts.png" alt="Terminal com o erro de execução de scripts, o comando Set-ExecutionPolicy e a ativação funcionando em seguida" width="1000">

- ① O erro: o Windows bloqueou o script de ativação.
- ② O comando que libera scripts para o seu usuário: `Set-ExecutionPolicy -Scope CurrentUser RemoteSigned`
- ③ O PowerShell pede confirmação: digite **S** (ou **Y**, se o Windows estiver em inglês) e aperte `Enter`.
- ④ Na nova tentativa, a ativação funciona e o `(.venv)` aparece.

> 💡 **Importante:** a ativação vale só para **aquela janela** do terminal. Sempre que abrir um terminal novo para trabalhar no curso, entre na pasta e ative o ambiente de novo. Para desativar, digite `deactivate`.

## 1.7 Instalando as bibliotecas com o pip

Com o ambiente **ativado** (o `(.venv)` aparecendo no terminal), instale as bibliotecas do curso. O arquivo `requirements.txt` contém a **lista de dependências** do curso: o `pip` lê essa lista e instala tudo de uma vez.

⌨️
```
pip install -r requirements.txt
```

O `-r` significa "leia a lista deste arquivo". O pip vai baixar as bibliotecas e mais algumas dezenas de outras de que elas dependem. Isso leva de 1 a 5 minutos, dependendo da sua internet.

<img src="imagens/terminal-pip-install.png" alt="Terminal com o pip install -r requirements.txt e a mensagem Successfully installed no final" width="1060">

- ① O comando de instalação (repare no `(.venv)` antes dele).
- ② O pip mostra cada biblioteca que encontra e baixa.
- ③ Quando termina, a última mensagem começa com **`Successfully installed`**, seguida do nome de todos os pacotes instalados.

Estas são as bibliotecas usadas nos Módulos 01 a 05, e onde cada uma entra:

| Biblioteca | Para que serve | Módulos |
|---|---|---|
| `notebook` | O Jupyter Notebook, onde escrevemos e executamos o código | 01 em diante |
| `requests` | Faz pedidos HTTP: baixa páginas e dados da internet | 02 em diante |
| `parsel` | Encontra os dados dentro do HTML, com XPath e seletores CSS | 03 em diante |
| `lxml` | Instalada junto com o `parsel`: é ela que lê o HTML por trás dele | 03 em diante |

O restante do que os módulos usam (`json`, `pathlib`, `time`, `urllib`) já vem com o Python. Os Módulos 06 em diante vão precisar de mais algumas bibliotecas, como o `pandas`. Quando eles forem publicados, essas bibliotecas entram no `requirements.txt`: basta baixar a versão nova do curso e rodar o `pip install -r requirements.txt` de novo.

> ⚠️ Se aparecer `ERROR` em vermelho no final, leia a última linha da mensagem e consulte a seção **Problemas comuns**, no final deste módulo. Já as mensagens que começam com `[notice]` (por exemplo, *A new release of pip is available*) são apenas avisos e podem ser ignoradas.

### Glossário

| Termo | Significado |
|---|---|
| **Terminal** | Programa para operar o computador por comandos digitados |
| **Prompt** | O começo da linha do terminal, que mostra a pasta atual |
| **Python** | A linguagem de programação e o programa que executa o código (o interpretador) |
| **Biblioteca** (ou pacote) | Código pronto que adiciona novas funções ao Python |
| **pip** | O instalador de bibliotecas do Python |
| **Ambiente virtual** (venv) | Pasta que isola as bibliotecas de um projeto |
| **Notebook** (`.ipynb`) | Documento com células de código e de texto |
| **Kernel** | O Python que executa as células e guarda as variáveis na memória |
| **Markdown** | Forma simples de formatar textos usando símbolos |
| **Traceback** | A mensagem de erro do Python |
